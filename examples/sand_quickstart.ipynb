# SAND quickstart

Load the bundled shop sample, inspect a table, join sales to customers, and chart revenue by region.

Run from the repo root after `./scripts/bootstrap.sh` and `pip install -e .`.

```python
from sand.jupyter import load
from pathlib import Path

sample_dir = Path("src/sand/samples")
ds = load(
    [sample_dir / "sales.csv", sample_dir / "customers.csv", sample_dir / "products.csv"],
    dataset_id="shop_nb",
)
ds.tables
```

```python
ds.profile("sales").head()
```

```python
enriched = ds.join(
    "sales",
    "customers",
    on="cust_id=id",
    how="left",
    as_table="sales_enriched",
)
enriched.head()
```

```python
fig = ds.chart(
    sql="SELECT region, SUM(amount) AS total FROM sales_enriched GROUP BY region ORDER BY total DESC",
    chart_type="bar",
    title="Revenue by region",
)
fig
```
